```python
import sys
import os

sys.path.insert(0, os.path.abspath("../.."))

import numpy as np
from nova.models import Sequential
from nova.models import Sequential
import pandas as pd
import nova.preprocessing as dp
from nova.layers import Dense
from nova.activations import Sigmoid, ReLu, Tanh, Softmax
from nova.losses import BinaryCrossEntropy, CategoricalCrossEntropy
from sklearn.metrics import f1_score


df = pd.read_csv("C:\\Users\\Luca\\Desktop\\Personal\\Nova\\datasets\\datasets\\AgenticAI_Leadership_Dataset_v1.csv")

train_set, val_set, test_set = dp.train_val_test_split(df)

X_train = train_set.drop(columns=["Adoption_Success_Level"])
y_train = train_set["Adoption_Success_Level"]
X_test = test_set.drop(columns=["Adoption_Success_Level"])
y_test = test_set["Adoption_Success_Level"]
X_val = val_set.drop(columns=["Adoption_Success_Level"])
y_val = val_set["Adoption_Success_Level"]

#Preprocessing

preparer = dp.DataFramePreparer()

preparer.fit(X_train)

X_train_prepared = preparer.transform(X_train)
X_test_prepared = preparer.transform(X_test)
X_val_prepared = preparer.transform(X_val)

#Modelo

model = Sequential()

#Layers

model.add(Dense(input_dim=X_train_prepared.shape[1], output_dim=10))
model.add(ReLu())
model.add(Dense(input_dim=10, output_dim=9))
model.add(ReLu())
model.add(Dense(input_dim=9, output_dim=8))
model.add(ReLu())
model.add(Dense(input_dim=8, output_dim=7))
model.add(ReLu())
model.add(Dense(input_dim=7, output_dim = 6))
model.add(ReLu())
model.add(Dense(input_dim=6, output_dim = 5))
model.add(ReLu())
model.add(Dense(input_dim=5, output_dim = 4))
model.add(ReLu())
model.add(Dense(input_dim=4, output_dim = 3))
model.add(ReLu())
model.add(Dense(input_dim=3, output_dim = 2))
model.add(ReLu())
model.add(Dense(input_dim=2, output_dim=3))
model.add(Softmax())

#Loss

y_train = y_train.map({"Low": 0, "Medium": 1, "High":2})
y_val = y_val.map({"Low": 0, "Medium": 1, "High":2})
y_test = y_test.map({"Low": 0, "Medium": 1, "High":2})

y_train_oh = np.eye(3)[y_train.to_numpy()]
y_val_oh   = np.eye(3)[y_val.to_numpy()]
y_test_oh  = np.eye(3)[y_test.to_numpy()]

model.compile(loss=CategoricalCrossEntropy())

model.fit(X_train_prepared.to_numpy(), y_train_oh, iterations_num=10000)

y_pred_test = model.forward(X_test_prepared.to_numpy())
```

```python
y_pred_classes = np.argmax(y_pred_test, axis=1)
y_true_classes = np.argmax(y_test_oh, axis=1)

print(f"F1 score: {f1_score(y_true_classes, y_pred_classes, average='weighted')}")
```
